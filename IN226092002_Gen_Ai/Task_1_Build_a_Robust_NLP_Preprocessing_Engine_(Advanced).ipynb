## Task 1: Conceptual Understanding

**Q1. What is the difference between "Love" and "love" in NLP?**

In NLP, "Love" and "love" are treated as different tokens because computers compare text letter by letter. They don't recognize them as the same word. That's why lowercasing is an important preprocessing step — it ensures the same word is always represented the same way.

**Q2. What happens if stopwords are not removed?**

If stopwords are not removed, the model might treat common meaningless words like "the", "is", "a" as important features, which reduces accuracy and slows down training unnecessarily.

**Q3. Two real-world scenarios where removing stopwords can be harmful:**

1. In sentiment analysis, the sentence "The food in the restaurant is not good" becomes "food restaurant good" after removing stopwords — completely flipping the meaning from negative to positive.
2. In a content filtering system, "This movie is not for people whose age is under 18" loses its restriction meaning entirely after stopword removal, which could be dangerous.

**Q4. Difference between Stemming and Lemmatization:**

Stemming crudely chops off word endings without checking if the result is a real word (e.g. "running" → "runn"). Lemmatization uses vocabulary and grammar rules to return a proper dictionary word (e.g. "running" → "run"). Lemmatization is slower but more accurate.

##Task 2:Build Advanced Preprocessing Function

```python
import re

def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'http\S+|www\S+', '', text)  # remove URLs
    text = re.sub(r'\S+@\S+', '', text)          # remove emails
    text = re.sub(r'\d+', '', text)              # remove numbers
    text = re.sub(r'[^\w\s]', '', text)          # remove punctuation
    text = re.sub(r'(.)\1{2,}', r'\1', text)     # handle repeated characters
    text = re.sub(r'\s+', ' ', text).strip()     # remove extra spaces
    tokens = text.split()
    tokens = [word for word in tokens if len(word) > 2 or word in ["no", "not"]]
    return tokens
```

##Task 3: Stress Testing

```python
sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

for s in sentences:
    tokens = preprocess_text(s)
    print("Original:", s)
    print("Tokens:", tokens)
    print("Clean Sentence:", ' '.join(tokens))
    print()
```

##Task 4: Token Analytics

```python
for s in sentences:
    tokens = preprocess_text(s)
    print(f"Sentence: {s}")
    print(f"Total tokens: {len(tokens)}")
    print(f"Unique tokens: {len(set(tokens))}")
    print(f"Avg token length: {round(sum(len(w) for w in tokens)/len(tokens), 2) if tokens else 0}")
    print()
```

##Task 5: Frequency Analysis

```python
from collections import Counter
all_tokens = [token for s in sentences for token in preprocess_text(s)]
freq = Counter(all_tokens)
print("Top 10 most frequent:", freq.most_common(10))
print("Top 5 least frequent:", freq.most_common()[:-6:-1])
```

##Task 6: Full Pipeline

```python
def full_pipeline(text_list):
    tokens = [preprocess_text(s) for s in text_list]
    clean_sentences = [' '.join(t) for t in tokens]
    return {"tokens": tokens, "clean_sentences": clean_sentences}

print(full_pipeline(sentences))
```
